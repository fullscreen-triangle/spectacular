# spectacular in Jupyter

Linked charts of one table: brush a range in any chart and every other chart filters. The charts are *described* in JSON, not coded, so a program can write the description as easily as a person.

Needs `pip install "spectacular-charts[notebook]"` (or, from the checkout, `npm run build` then `maturin develop --release` in `python/`).

```python
import pandas as pd
import spectacular as sp  # also turns on the %%spectacular cell magic

df = pd.read_csv("../lorenz.csv")  # 29,900 states of the Lorenz system
df.head()
```

```python
%%spectacular df --as w
{"title": "Lorenz attractor",
 "labels": {"t": "time", "speed": "|dx/dt|"},
 "columns": 2,
 "charts": [
   {"x": "x", "y": "y", "z": "z", "color": "speed", "title": "Phase space"},
   {"x-axis": "x", "y-axis": "z", "color": "lobe", "palette": ["steelblue", "darkorange"]},
   {"x": "t", "bins": 150, "wide": true},
   {"x": "speed", "color": "seagreen"},
   {"x": "lobe"}
 ]}
```

Brush a range of time or speed above, or click a lobe, then run the cells below. The selection comes back in data terms, and as the rows themselves.

```python
w.selection
```

```python
w.selected_data().describe()
```

The same charts leave the notebook as an offline page, as an SVG of what is on screen now (brushes included), or as a figure at print size. `render` draws in headless Chromium: `pip install "spectacular-charts[render]"`, then `python -m playwright install chromium`.

```python
description = {"title": "Lorenz attractor", "theme": "paper",
               "figure": {"format": "pdf", "width": 180, "unit": "mm"},
               "charts": [{"x": "x", "y": "z", "color": "lobe"}, {"x": "speed"}]}
sp.save_html(description, "lorenz.html", data=df)
w.save_svg("lorenz-brushed.svg")
sp.render(description, "lorenz.pdf", data=df)  # needs spectacular-charts[render]
```
