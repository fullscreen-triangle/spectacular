# spectacular in Jupyter

Linked views of one table: brush a range in any view and every other view filters. The spec is plain data, so a program can build it as easily as a person.

Needs `pip install -e "python[notebook]"` from the spectacular checkout, after `npm run build`.

```python
import pandas as pd
import spectacular as sp

df = pd.read_csv("../lorenz.csv")  # 29,900 states of the Lorenz system
df.head()
```

```python
spec = {
    "title": "Lorenz attractor",
    "fields": {"t": {"title": "time"}, "speed": {"title": "|dx/dt|"}},
    "layout": {"columns": 2},
    "views": [
        {"id": "attractor", "type": "scatter3d", "title": "Phase space", "x": "x", "y": "y", "z": "z", "color": "speed"},
        {"id": "xz", "type": "scatter", "title": "x-z projection", "x": "x", "y": "z", "color": "lobe"},
        {"id": "time", "type": "histogram", "title": "Time", "field": "t", "bins": 150, "span": 2},
        {"id": "speed", "type": "histogram", "title": "Speed", "field": "speed", "bins": 80},
        {"id": "lobe", "type": "bars", "title": "Lobe", "field": "lobe"},
    ],
}
w = sp.view(spec, data=df)
w
```

Brush a range of time or speed above, or click a lobe, then run the cells below. The selection comes back in data terms, and as the rows themselves.

```python
w.selection
```

```python
w.selected_data().describe()
```

The same views leave the notebook as an offline page, as an SVG of what is on screen now (brushes included), or as a figure at print size. `render` runs the `spectacular` command, so it needs Node.

```python
sp.save_html(spec, "lorenz.html", data=df)
w.save_svg("lorenz-brushed.svg")
sp.render({**spec, "theme": "paper", "output": {"figure": {"format": "pdf", "width": 180, "unit": "mm"}}}, "lorenz.pdf", data=df)
```
